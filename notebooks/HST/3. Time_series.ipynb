```python
from astropy.io import fits
from astropy.stats import poisson_conf_interval
import matplotlib.pyplot as plt
import matplotlib.pylab as pylab
import numpy as np
from rocky_worlds_utils.hst import time_series, stis_analysis, tools

%matplotlib inline
pylab.rcParams['figure.figsize'] = 15.0,9.0
pylab.rcParams['font.size'] = 18
```

In the notebook \#2, we custom extracted data from two STIS/FUV-MAMA exposures for the target GJ 3929. Now, we shall analyze its spectral time series. For this example, we are interested in the stellar fluxes coming from the Lyman-alpha line.

The first step of this analysis is to take advantage of the `TIME-TAG` mode, which assigns a timestamp to every count that hits the detector and allows for a temporal resolution down to that of the electronics! We do not need all of that. In fact, here we shall work with a coarser temporal resolution that is more relevant for the analysis of stellar flares.

We shall use the `stis_analysis.timetag_split()` function to break down a full exposure into sub-exposures. This could take a while because in the end we need to process many subexposures.

```python
time_resolution = 60  # unit of seconds
prefix = './example_data/'
output_dir = './example_data/timeseries/'
datasets = ['ofkb08020', 'ofkb08030']

for dataset in datasets:
    stis_analysis.timetag_split(dataset, 
                                prefix, 
                                output_dir=output_dir, 
                                temporal_resolution=time_resolution, 
                                overwrite=False)
```

The code above should have generated `*_ts_x1d.fits` files inside the output folder (`ts` stands for time-series). Inside these new files, there are many extensions that correspond to each subexposure. We can use them to generate a light curve of the red wing of the Lyman-alpha line.

First, we need to find out which wavelength range over which we want to integrate the flux densities. Let us visually inspect the Lyman-alpha profile.

```python
y_plot_scale = 1E-14
for i, dataset in enumerate(datasets):
    spec = fits.getdata(prefix + dataset + '_x1d.fits')
    header = fits.getheader(prefix + dataset + '_x1d.fits', ext=0)
    wl = spec['WAVELENGTH'].ravel()
    f = spec['FLUX'].ravel()
    err = spec['ERROR'].ravel()
    plt.errorbar(wl, f / y_plot_scale, yerr=err / y_plot_scale, fmt='o', mec='white', ls='-', alpha=0.3, color='C'+str(i))
    plt.plot(wl, f / y_plot_scale, lw='2', color='C'+str(i), label=dataset)
plt.xlabel(r'Wavelength [${\rm \AA}$]')
plt.ylabel(r'Flux density [$\times 10^{%i}$ erg s$^{-1}$ cm$^{-2}$ ${\rm \AA}^{-1}$]' % int((np.log10(y_plot_scale))))
plt.xlim(1210, 1220)
plt.ylim(-0.1, 2.1)

# From a visual inspection, it seems that most of the stellar Lya flux is located within 1215.7 and 1217.5 Angstrom
plt.axvspan(xmin=1215.7, xmax=1217.5, color='C3', alpha=0.1)
# There is also some emission flux near 1218, but that is the OV line and we are not interested in it for this example.

plt.legend()
plt.tight_layout()
```

From a first-pass visual inspection, it seems that most of the stellar Lya flux is located within 1215.9 and 1217.5 Angstrom. So, we shall generate a light curve within this range.

```python
wavelength_range = (1215.7, 1217.5)

for dataset in datasets:
    time, flux, flux_error = time_series.generate_light_curve(
        dataset,
        prefix=output_dir,
        wavelength_range=wavelength_range
    )
    plt.errorbar(time, flux, yerr=flux_error, color='k', fmt='o')
plt.xlabel('Time (MJD)')
plt.ylabel(r'Flux (erg s$^{-1}$ cm$^{-2}$)')
```

We should, however, be careful with the geocoronal contamination, which is treated as background in STIS/FUV-MAMA data. This contamination tends to increase the noise of Lya spectra near the rest wavelength of the Lya line. In order to check where the geocoronal contamination is the strongest, it is useful to compare the `BACKGROUND` count rate with the `NET` (=`GROSS`-`BACKGROUND`) count rate.

```python
for i, dataset in enumerate(datasets):
    spec = fits.getdata(prefix + dataset + '_x1d.fits')
    header = fits.getheader(prefix + dataset + '_x1d.fits', ext=0)
    wl = spec['WAVELENGTH'].ravel()
    gross = spec['BACKGROUND'].ravel()
    net = spec['NET'].ravel()
    plt.plot(wl, gross, lw='2', color='C'+str(i), label=dataset)
    plt.plot(wl, net, ls='--', lw='2', color='C'+str(i))
plt.xlabel(r'Wavelength [${\rm \AA}$]')
plt.ylabel(r'Count rate (e$^{-}$ s$^{-1}$)')
plt.xlim(1210, 1220)

# The geocoronal contamination seems to be the strongest within this range
plt.axvspan(xmin=1215.2, xmax=1215.9, color='k', alpha=0.1)

plt.legend()
plt.tight_layout()
```

We can see that the background is stronger than the stellar Lya red wing between wavelengths 1215.2 and 1215.9 Angstrom. We would benefit from avoiding integrating fluxes within this range to minimize the noise in the light curve. Let's see how the blue and red wing light curves look.

```python
for dataset in datasets:
    wavelength_range = (1215.9, 1217.5)  # Red wing
    time, flux, flux_error = time_series.generate_light_curve(
        dataset,
        prefix=output_dir,
        wavelength_range=wavelength_range
    )
    plt.errorbar(time, flux, yerr=flux_error, color='C3', fmt='o')

    wavelength_range = (1214.0, 1215.2)  # Blue wing
    time, flux, flux_error = time_series.generate_light_curve(
        dataset,
        prefix=output_dir,
        wavelength_range=wavelength_range
    )
    plt.errorbar(time, flux, yerr=flux_error, color='C0', fmt='o')

plt.xlabel('Time (MJD)')
plt.ylabel(r'Flux (erg s$^{-1}$ cm$^{-2}$)')
```

You can explore other temporal resolutions that are the most convenient for your science case by setting different values of `time_resolution`. You can also set `overwrite=True` when running `stis_analysis.timetag_split()` to avoid creating multiple files.

For now, we shall stick to the 60-s temporal resolution and we will create a High-Level Science Product, or HLSP, that contains light curves of spectral features of interest in a given dataset. There is a handy function for that: `time_series.generate_lc_hlsp()`.

This function accepts multiple wavelength ranges, corresponding to multiple spectral features. In the HLSP file, each data extension will contain the light curve of each spectral feature.

```python
wl_range = np.array([[1214, 1215.2], 
                     [1215.9, 1218]])
feature_names = ('LyA blue wing', 
                 'LyA red wing')

# A data source DOI is required; it can be obtained from MAST 
# for a subset of HST observations. In the case of the RWDDT data
# of GJ 3929, we have already generated a DOI.
data_doi = "10.17909/8pvd-9663"

# It is good practice to version HLSPs. It must be passed as a string.
version = "1.0"

time_series.generate_lc_hlsp(datasets, 
                             prefix=output_dir, 
                             wavelength_ranges=wl_range, 
                             source_doi=data_doi, 
                             feature_names=feature_names, 
                             output_dir=output_dir,
                             version=version)
```

And a light curve HLSP has now been generated! The filename is automatically generated too: `hlsp_rocky-worlds_hst_stis_gj3929_g140m_v1.0_lc.fits`. The headers contain important metadata, such as the observation dates in MJD, the time standard and proposal ID.

```python
hlsp_filename = 'hlsp_rocky-worlds_hst_stis_gj3929_g140m_v1.0_lc.fits'
header_0 = fits.getheader(output_dir + hlsp_filename, ext=0)
print(repr(header_0))
```

Note that the data extensions have their own metadata too, including the wavelength ranges, measurement units, description of the data and average sensitivity of the wavelength range.

```python
header_1 = fits.getheader(output_dir + 'hlsp_rocky-worlds_hst_stis_gj3929_g140m_v1.0_lc.fits', ext=1)
print(repr(header_1))
```

Using data from the light curve HLSP, we can combine fluxes of non-contiguous spectral features. For example, let us combine the blue wing and red wing fluxes to produce a higher-SNR light curve. Care has to be taken with the uncertainties because we are working with low count rates in the case of GJ 3929, so we need to propagate them using Poisson statistics.

There is a handy function `tools.coadd_flux()` that does that job for us.

```python
blue_lya_lc = fits.getdata(output_dir + hlsp_filename, ext=1)
red_lya_lc = fits.getdata(output_dir + hlsp_filename, ext=2)

# Fetch the relevant data
time = blue_lya_lc['TIME']
sub_exp_time = blue_lya_lc['SUBEXPTIME']

# The non-contiguous data should be arranged in numpy.ndarrays
# where the first dimension (or row) should be correspondent
# to the number of wavelength ranges
lya_f = np.array([blue_lya_lc['FLUX'], red_lya_lc['FLUX']])
lya_gross = np.array([blue_lya_lc['GROSS'], red_lya_lc['GROSS']])
lya_sens = np.array([blue_lya_lc['SENSITIVITY'], red_lya_lc['SENSITIVITY']]) 

total_f, total_f_err = tools.coadd_flux(lya_f, lya_gross, lya_sens, sub_exp_time)

# Finally, we plot the light curve of the combined red and blue
# Lya wings
plt.errorbar(time, total_f, yerr=total_f_err, color='k', fmt='o')

plt.xlabel('Time (MJD)')
plt.ylabel(r'Flux (erg s$^{-1}$ cm$^{-2}$)')
```
