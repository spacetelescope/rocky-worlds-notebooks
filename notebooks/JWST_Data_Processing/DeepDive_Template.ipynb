### 1.1 Auto-tuning your Stage 1 settings (optional)

If `run_optimizers` is `True`, this uses Eureka!'s Stage 1 optimizer to automatically sweep `jump_rejection_threshold`, `skip_firstframe`, `skip_lastframe`, `skip_rscd`, `rscd_group_skip1`, and `rscd_group_skip` (only processing the first file segment, to keep this fast) and recommend values based on the resulting Stage 3/4 noise metrics, instead of requiring you to manually guess-and-check these settings as described in the `README_DeepDive.md` file. We run the optimizer first, before setting up this target's Stage 1 ECF below, since it works from the raw uncalibrated data and uses its own generic optimizer ECFs. The recommended values are automatically applied when Stage 1 is run for real in Section 1.3. Set `run_optimizers` to `False` above to skip this and use the hardcoded default settings instead.

```python
best_s1 = {}
if run_optimizers:
    s1opt_ecf_contents = f"""# Eureka! Control File for Optimizing Stage 1

# Optimizer Documentation: https://eurekadocs.readthedocs.io/en/latest/ecf.html#stages-1-3-4-optimizer

scaling_MAD_spec    0
scaling_MAD_white   1

# List of Stage 1 parameters to optimize (explicit here rather than relying on
# built-in defaults, so the exact sweep is easy to review/adjust).
params_to_optimize_s1           ['jump_rejection_threshold', 'skip_lastframe', 'rscd_group_skip1', 'rscd_group_skip', 'skip_rscd']
sweep_jump_rejection_threshold  np.arange(4,15)
sweep_skip_lastframe            [True, False]
sweep_rscd_group_skip1          np.arange(0,6)
sweep_rscd_group_skip           np.arange(0,6)
sweep_skip_rscd                 [False, True]

# Diagnostics
isplots_S1opt       1    # Generate none (0), few (1), some (3), or many (5) figures
delete_intermediate True    # Delete each sweep step's intermediate outputs to save disk space
delete_final        True    # This notebook does its own confirmatory Stage 1 run below
verbose             True
hide_plots          True

# Project directory
topdir              {topdir}

# Directories relative to topdir
inputdir            Uncalibrated
outputdir           {analysis_name}/DeepDive/Stage1opt
"""

    # This will save the ECF as a file that the optimizer can read-in
    with open('./S1opt_rwddt_optimizer.ecf', 'w') as f:
        f.write(s1opt_ecf_contents)

    # The optimizer builds internal Stage 1/2/3/4 Meta objects by looking up
    # S{stage}_rwddt_optimizer.ecf. Using this generic 'rwddt_optimizer' eventlabel
    # (instead of this target's real eventlabel) means those static ecf files always
    # exist here, so this cell won't crash when the target's own S2/S3/S4 ecfs
    # haven't been written yet, and the optimizer uses single aperture+annulus
    # settings in the S3/S4 static ecfs instead of the target's full [lo,hi,step] grid.
    s1opt_meta, s1opt_history, best_s1 = s1opt.wrapper('rwddt_optimizer', ecf_path='.',
                                                        initial_run=True, final_run=False)
```

### 1.2 Setting up the Stage 1 ECF

A good starting point for your deep-dive analysis has already been populated within the template Stage 1 ECF file, which we load in and look at here. The Stage 1 optimizer above (Section 1.1) recommends values for a handful of these settings; when `run_optimizers` is `True`, those recommendations are written straight into this ECF through the `{s1_params[...]}` placeholders, falling back to the hardcoded defaults in the `s1_params` dictionary when `run_optimizers` is `False`. To override any of them, edit its value in `s1_params` or replace its `{...}` placeholder in the ECF with a literal value.

Parameters that might be worth varying are as documented in the `README_DeepDive.md` file.

```python
# The Stage 1 optimizer tunes these settings. With run_optimizers = True they are inherited from
# its `best_s1` recommendation; otherwise the hardcoded fallbacks below are used. Override one by
# editing its value here or by replacing its {...} placeholder in the ECF with a literal value.
s1_params = {
    'skip_lastframe':           False,
    'skip_rscd':                False,
    'jump_rejection_threshold': 8.0,
    'rscd_group_skip1':         None,
    'rscd_group_skip':          None,
}
if run_optimizers and best_s1:
    s1_params.update({key: (val.item() if hasattr(val, 'item') else val)
                      for key, val in best_s1.items() if key in s1_params})

s1_ecf_contents = f"""# Eureka! Control File for Stage 1: Detector Processing

# Stage 1 Documentation: https://eurekadocs.readthedocs.io/en/latest/ecf.html#stage-1

suffix                    uncal
pmap                      1535    # Setting a fixed pmap value to ensure consistency in reductions

maximum_cores             'half'    # Options are 'none', 'quarter', 'half', 'all', or any integer

# Pipeline stages
skip_emicorr              False    #### LOW PRIORITY: If observations use SUB64 or SUB128 subarrays, turning this to True will leave behind EMI artifacts, but may reduce noise in other ways
skip_saturation           False
skip_lastframe            {s1_params['skip_lastframe']}    #### HIGH PRIORITY: If observations have a small number of groups, turning this to True will increase the number of frames used (reducing photon noise), but at the cost of adding a known-noisy frame
skip_reset                False
skip_linearity            False
skip_rscd                 {s1_params['skip_rscd']}    #### MEDIUM PRIORITY: If observations have a small number of groups, turning this to True will increase the number of frames used (reducing photon noise), but at the cost of adding known-noisy frames
skip_dark_current         True    #### LOW PRIORITY: Turning this to False will subtract a dark current reference file, which may reduce noise in some cases but can also add noise if the reference file has artifacts
skip_refpix               False
skip_jump                 False
skip_clean_flicker_noise  True    #### LOW PRIORITY: MIRI has little "flicker" or 1/f noise, so this step is generally not necessary for MIRI data.

# Pipeline stages parameters
jump_rejection_threshold  {s1_params['jump_rejection_threshold']}    #### MEDIUM PRIORITY: Increasing this value will reduce the number of good frames marked as cosmic rays, which will reduce photon noise but may also allow more cosmic rays to remain in the data. Meanwhile, decreasing it will increase photon noise but may also remove more cosmic rays.
rscd_group_skip1          {s1_params['rscd_group_skip1']}    #### Number of groups to flag as DO_NOT_USE via the RSCD step in the FIRST integration only. Requires the eureka.optimizer/miri_firstframe RSCD machinery; has no effect on branches without it (e.g. tjb_rwddt).
rscd_group_skip           {s1_params['rscd_group_skip']}    #### Number of groups to flag as DO_NOT_USE via the RSCD step in all LATER integrations. Same branch requirement as rscd_group_skip1.

# Diagnostics
isplots_S1                3
nplots                    5
hide_plots                True
verbose                   True

# Project directory
topdir              {topdir}

# Directories relative to topdir
inputdir            Uncalibrated
outputdir           {analysis_name}/DeepDive/Stage1
"""

# This will save the ECF as a file that the next cell can read-in
with open(f'./S1_{eventlabel}.ecf', 'w') as f:
    f.write(s1_ecf_contents)
```

## 3. Stage 3 - Pixels to Lightcurve

Stage 3 performs background subtraction and optimal spectral extraction. This will generate time series of 1D spectra. Because Eureka!'s optimizer considers Stage 3 and Stage 4 settings together, we (optionally) run the Stage 3/4 optimizer first, immediately after Stage 2, and then set up this target's Stage 3 and Stage 4 ECFs below. If `run_optimizers` is `True`, the optimizer's recommendation is used to narrow the aperture/annulus grid (and set `bg_thresh`/`dqmask`/`sigma`/`box_width`) that Stage 3 and Stage 4 actually run over, instead of the full grid written out in those ECFs.

### 3.1 Auto-optimizing your Stage 3/4 settings (optional)

If `run_optimizers` is `True`, this uses Eureka!'s Stage 3/4 optimizer to quickly sweep `photap`, `skyin`, `skywidth`, `bg_thresh`, and `dqmask` (Stage 3) and `sigma` and `box_width` (Stage 4) using a fast pixel-level/white-light-curve MAD fitness metric (no Stage 5 fitting required). Because this runs immediately after Stage 2 - before Stage 3 or Stage 4 have actually been run - its recommendation is used to narrow the `photap`/`skyin` grid, hold `skywidth` fixed at its recommended value (since annulus width is normally not very important), and set the `bg_thresh`/`dqmask`/`sigma`/`box_width` values used when Stage 3 and Stage 4 are run for real below, instead of running the full grid written out below. This is where the time savings come from, since exploring the full grid by hand can otherwise require dozens of Stage 3/4 combinations (and eventually Stage 5 fits). Set `run_optimizers` to `False` above to skip this and use the full grid and hardcoded defaults written out below, as before.

```python
best = {}
if run_optimizers:
    s3opt_ecf_contents = f"""# Eureka! Control File for Optimizing Stages 3 and 4

# Optimizer Documentation: https://eurekadocs.readthedocs.io/en/latest/ecf.html#stages-1-3-4-optimizer

scaling_MAD_spec    0
scaling_MAD_white   1

# List of Stage 3-4 parameters to optimize (explicit here rather than relying on
# built-in defaults, so the exact sweep is easy to review/adjust).
params_to_optimize_s3   ['photap', 'skyin', 'skywidth', 'bg_thresh', 'dqmask']
sweep_photap            np.arange(4,12)
sweep_skyin             np.arange(14,30,4)
sweep_skywidth          np.arange(10,40,10)
sweep_bg_thresh         np.arange(3.0,5.5,0.5)
sweep_dqmask            [True, False]

params_to_optimize_s4   ['sigma', 'box_width']
sweep_sigma             np.arange(4.,6.,0.5)
sweep_box_width         np.arange(11,52,10)

# Set to True if you want to optimize based only on the first file segment (much faster, but less robust)
firstSegOnly_S3     True

# Diagnostics
isplots_S3opt       1
delete_intermediate True
delete_final        True
verbose             True
hide_plots          True

# Project directory
topdir              {topdir}

# Directories relative to topdir
inputdir            {s2_meta.outputdir_raw}
outputdir           {analysis_name}/DeepDive/Stage3opt
"""

    # This will save the ECF as a file that the optimizer can read-in
    with open('./S3opt_rwddt_optimizer.ecf', 'w') as f:
        f.write(s3opt_ecf_contents)

    # The optimizer builds internal Stage 3/4 Meta objects by looking up
    # S3_rwddt_optimizer.ecf and S4_rwddt_optimizer.ecf. Those static ecf files keep
    # photap/skyin/skywidth as single scalars (not the [lo,hi,step] grid used for the
    # real target below), so every sweep iteration processes one aperture combination
    # instead of the full 48-combo grid. Using the generic 'rwddt_optimizer' eventlabel
    # (instead of the target's real eventlabel) is what points the optimizer at those
    # fast static files.
    s3opt_meta, s3opt_history, best = s3opt.wrapper('rwddt_optimizer', ecf_path='.',
                                                     initial_run=True, final_run=False)

    print('Stage 3/4 optimizer recommendation:', best)
```

### 3.2 Setting up the Stage 3 ECF

Parameters that might be worth varying are documented in the `README_DeepDive.md` file. When `run_optimizers` is `True`, the `photap`/`skyin`/`skywidth`/`bg_thresh`/`dqmask` values are inherited from the Stage 3/4 optimizer's recommendation (assembled in the `s3_params` dictionary and written into the ECF via the `{...}` placeholders); when it is `False`, the hardcoded defaults in `s3_params` are used instead. To override any value, edit it in `s3_params` or replace its `{...}` placeholder in the ECF with a literal value.

By default, the most recently created Stage 2 output located in `s3_meta.inputdir` will be used as the input for your Stage 3. If you instead want to use a specific Stage 2 output, you'll need to update your `inputdir` setting to point to the specific Stage 2 output folder you want to work on (e.g., `'Stage2/S2_2025-02-28_trappist1b_eclipse4_run1').

```python
# Optimizer-tunable Stage 3 settings (aperture/annulus grid + background/DQ handling). With
# run_optimizers = True, photap/skyin become a small grid centered on the optimizer's `best`
# recommendation, skywidth is held fixed at its recommended value, and bg_thresh/dqmask take
# their recommended values; otherwise the hardcoded defaults below are used. Override any value
# by editing it here or by replacing its {...} placeholder in the ECF with a literal value.
s3_params = {
    'photap':    [4, 11, 1],
    'skyin':     [14, 26, 4],
    'skywidth':  [10, 30, 10],
    'bg_thresh': [5, 5],
    'dqmask':    True,
}
if run_optimizers and 'photap' in best:
    photap_ctr = int(best['photap'])
    skyin_lo = max(photap_ctr, int(best['skyin']) - 4)
    s3_params['photap'] = [max(4, photap_ctr - 2), photap_ctr + 2, 1]
    s3_params['skyin'] = [skyin_lo, skyin_lo + 8, 4]
    s3_params['skywidth'] = int(best['skywidth'])  # held fixed; annulus width is normally not very important
    s3_params['bg_thresh'] = [float(best['bg_thresh']), float(best['bg_thresh'])]
    s3_params['dqmask'] = bool(best['dqmask'])

s3_ecf_contents = f"""# Eureka! Control File for Stage 3: Data Reduction

# Stage 3 Documentation: https://eurekadocs.readthedocs.io/en/latest/ecf.html#stage-3

ncpu            16
nfiles          200
max_memory      1.5
indep_batches   False    # Independently treat each batch of files? Strongly recommended to leave this as False unless you have a clear reason to set it to True.
suffix          calints

calibrated_spectra  False    # Set True to generate flux-calibrated spectra/photometry in mJy
                             # Set False to convert to electrons
pmap            1535

# Subarray region of interest
ywindow         None    # Vertical axis as seen in DS9
xwindow         None    # Horizontal axis as seen in DS9
dqmask          {s3_params['dqmask']}

# Background parameters
ff_outlier      True    # Set False to use only background region (recommended for deep transits)
                        # Set True to use full frame (works well for shallow transits/eclipses)
bg_thresh       {s3_params['bg_thresh']}
interp_method   linear    # Interpolate bad pixels. Options: None (if no interpolation should be performed), linear, nearest, cubic

# Centroiding parameters
centroid_method mgmc    # Method used for centroiding. Options: mgmc, fgc
ctr_guess		fits    # Initial guess of centroid position. If None, will first perform centroiding on whole frame (can sometimes fail)
ctr_cutout_size 10    # Cutoff size all around the centroid after the coarse centroid calculation or first centroid guess when using the mgmc method.
centroid_tech   com    # (mgmc method param) Technique used for centroiding. Options: com, 1dg, 2dg
gauss_frame     15    # (mgmc method param) Half-width away from second centroid guess to include in centroiding map for gaussian widths. Recommend ~15 for MIRI photometry.

# Photometric extraction parameters
phot_method     photutils    # photutils (aperture photometry using photutils), poet (aperture photometry using code from POET), or optimal (for optimal photometric extraction)
aperture_edge   exact    # center (pixel is included only if its center lies within the aperture), or exact (pixel is weighted by the fractional area that lies within the aperture)
aperture_shape  circle    # If phot_method is photutils or optimal: circle, ellipse, or rectangle. If phot_utils is poet: circle or hexagon. Used to set both the object aperture shape and the sky annulus shape
moving_centroid False    # Boolean: False if the aperture should stay fixed on the median centroid location (recommended), or True if the aperture should track the moving centroid
skip_apphot_bg  False    # Skips the background subtraction during the aperture photometry step

# With run_optimizers = True these are a small grid centered on the optimizer's recommendation; with run_optimizers = False they are the full default grid (only adjust if a first run suggests you need smaller or larger radii).
photap          {s3_params['photap']}    # Size of photometry aperture radius in pixels
skyin           {s3_params['skyin']}    # Inner sky annulus edge, in pixels
skywidth        {s3_params['skywidth']}    # Width of the sky annulus, in pixels

# Diagnostics
isplots_S3      3
nplots          3
hide_plots      False
save_output     True
save_fluxdata   False
verbose         True

# Project directory
topdir          {topdir}

# Directories relative to topdir
inputdir        {analysis_name}/DeepDive/Stage2
outputdir       {analysis_name}/DeepDive/Stage3
"""

# This will save the ECF as a file that the next cell can read-in
with open(f'./S3_{eventlabel}.ecf', 'w') as f:
    f.write(s3_ecf_contents)
```

### 3.4 Investigating the Stage 3 outputs

If there are issues with your analyses that you are having troubles working out, try re-running and increasing your `isplots_S3` setting to `5` to get more output plots that can sometimes be useful for debugging and/or increasing `nplots` to a larger number to investigate some of the repeated plots for more integrations or try setting `nplots` to `None` to make the repeated plots for all integrations (this will take a while and will make a *lot* of figures).

Generally speaking, the lower the Stage 3 MAD (median-absolute deviation) value, the better the overall reduction is since the S3 MAD is approximately what the overall noise level is in the lightcurve. This is not always the case, as sometimes lightcurves with higher MAD values have cleaner correlated noise properties that can be better decorrelated in Stage 5. But generally speaking, the lower the S3 MAD value the better so long as that isn't coming at the cost of large amounts of data points being discarded. A decent first approach is to try different Stage 3 settings until you find a "sweet-spot", and then if you want to double-check your work and conclusions, you can run Stages 4-5 on different Stage 3 outputs to confirm that your lower-MAD reduction resulted in a "better" final result.

## 4. Stage 4 - Removing time-series outliers

Stage 4 for photometric data just clips outliers with respect to a box-car smoothed version of the signal - this helps remove any remaining outliers while also trying to avoid removing astrophysical signals. Its ECF is set up in Section 4.1 below.

### 4.1 Setting up the Stage 4 ECF

Parameters that might be worth varying are documented in the `README_DeepDive.md` file. When `run_optimizers` is `True`, the `sigma`/`box_width` clipping settings are inherited from the Stage 3/4 optimizer's recommendation (assembled in the `s4_params` dictionary and written into the ECF via the `{...}` placeholders); otherwise the hardcoded defaults in `s4_params` are used. To override either, edit it in `s4_params` or replace its `{...}` placeholder in the ECF with a literal value.

By default, the most recently created Stage 3 output located in `s4_meta.inputdir` will be used as the input for your Stage 4. If you instead want to use a specific Stage 3 output, you'll need to update your `inputdir` setting to point to the specific Stage 3 output folder you want to work on (e.g., `'Stage3/S3_2025-02-28_trappist1b_eclipse4_run1'`).

```python
# The Stage 4 optimizer tunes these two clipping settings. With run_optimizers = True they are
# inherited from the optimizer's `best` recommendation; otherwise the hardcoded fallbacks below
# are used. Override one by editing its value here or by replacing its {...} placeholder in the ECF.
s4_params = {
    'sigma':     4.5,
    'box_width': 31,
}
if run_optimizers and 'sigma' in best:
    s4_params.update({key: (val.item() if hasattr(val, 'item') else val)
                      for key, val in best.items() if key in s4_params})

s4_ecf_contents = f"""# Eureka! Control File for Stage 4: Generate Lightcurves

# Stage 4 Documentation: https://eurekadocs.readthedocs.io/en/latest/ecf.html#stage-4

# Number of spectroscopic channels spread evenly over given wavelength np.arange
nspecchan       1
compute_white   False

allapers        True    # Run S4 on all of the apertures considered in S3? Otherwise will use newest output in the inputdir

# Parameters for sigma clipping
clip_binned     True    # Whether or not sigma clipping should be performed on the binned 1D time series
sigma           {s4_params['sigma']}    # The number of sigmas a point must be from the rolling median to be considered an outlier
box_width       {s4_params['box_width']}    # The width of the box-car filter (used to calculated the rolling median) in units of number of data points
maxiters        20    # The number of iterations of sigma clipping that should be performed.
boundary        fill    # Use 'fill' to extend the boundary values by the median of all data points (recommended), 'wrap' to use a periodic boundary, or 'extend' to use the first/last data points
fill_value      mask    # Either the string 'mask' to mask the outlier values (recommended), 'boxcar' to replace data with the mean from the box-car filter, or a constant float-type fill value.

# Project directory
topdir          {topdir}

# Directories relative to topdir
inputdir        {analysis_name}/DeepDive/Stage3
outputdir       {analysis_name}/DeepDive/Stage4
"""

# This will save the ECF as a file that the next cell can read-in
with open(f'./S4_{eventlabel}.ecf', 'w') as f:
    f.write(s4_ecf_contents)
```

### 4.2 Running Stage 4

Here we run the Eureka! Stage 4 pipeline using the settings we defined in Section 4.1, which already hold the optimizer's `sigma`/`box_width` recommendation when `run_optimizers` is `True`. For each aperture-annulus pair, this should take << 1 minute

```python
if process:
    # The Stage 4 ECF already holds the optimizer's sigma/box_width (or your defaults), so just run it
    spec, lc, s4_meta = s4.genlc(eventlabel)
else:
    # If not running S4, automatically grab the latest MetaData and SpecData
    temp_meta = MetaClass()
    temp_meta.eventlabel = eventlabel
    temp_meta.topdir = topdir
    temp_meta.inputdir = f'{topdir}/{analysis_name}/DeepDive/Stage4/'
    s4_meta, path, _ = findevent(temp_meta, 'S4')
    spec = xr.load_dataset(s4_meta.filename_S4_SpecData)
    lc = xr.load_dataset(s4_meta.filename_S4_LCData)
```
